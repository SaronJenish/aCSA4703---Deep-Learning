```python
# Common setup – run this cell first
import os, time, warnings, math
warnings.filterwarnings("ignore")
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             confusion_matrix, mean_absolute_error, mean_squared_error)
np.random.seed(42)
plt.rcParams["figure.figsize"] = (7, 4)
print("Setup complete")
```

## Experiment 22 – Independent Component Analysis and Component Visualization

```python
from scipy import signal
from sklearn.decomposition import FastICA, PCA
t=np.linspace(0,8,2000)
S=np.c_[np.sin(2*t),signal.square(2*np.pi*.5*t),signal.sawtooth(2*np.pi*.3*t)]
S=(S-S.mean(0))/S.std(0); A=np.array([[1,1,.5],[.5,2,1],[1.5,1,.5]])
Xmix=S@A.T
ica=FastICA(n_components=3,random_state=42, max_iter=1000); Sica=ica.fit_transform(Xmix)
Spca=PCA(n_components=3).fit_transform(Xmix)
fig,axs=plt.subplots(4,1,figsize=(12,8))
for ax,data,title in zip(axs,[S,Xmix,Sica,Spca],["Original sources","Mixed signals","ICA recovered","PCA components"]):
    ax.plot(data[:500]); ax.set_title(title)
plt.tight_layout(); plt.show()
print("Best absolute source correlations (ICA):")
print(np.round(np.max(np.abs(np.corrcoef(S.T,Sica.T)[:3,3:]),axis=1),3))
```
