## Experiment 8 – Two-Layer Neural Network with Backpropagation

```python
# Two-layer network trained on XOR; manual backpropagation
X=np.array([[0,0],[0,1],[1,0],[1,1]],float); y=np.array([[0],[1],[1],[0]],float)
rng=np.random.default_rng(4); W1=rng.normal(0,1,(2,4)); b1=np.zeros((1,4)); W2=rng.normal(0,1,(4,1)); b2=np.zeros((1,1))
sig=lambda z:1/(1+np.exp(-np.clip(z,-40,40)))
losses=[]; lr=.8
for epoch in range(10000):
    h=sig(X@W1+b1); out=sig(h@W2+b2)
    loss=np.mean((out-y)**2); losses.append(loss)
    do=2*(out-y)*out*(1-out)/len(X)
    dW2=h.T@do; db2=do.sum(0,keepdims=True)
    dh=(do@W2.T)*h*(1-h); dW1=X.T@dh; db1=dh.sum(0,keepdims=True)
    W2-=lr*dW2; b2-=lr*db2; W1-=lr*dW1; b1-=lr*db1
print("XOR predictions:\n",np.round(out,3))
plt.plot(losses); plt.xlabel("Epoch"); plt.ylabel("MSE"); plt.show()
# Decision regions
gx,gy=np.meshgrid(np.linspace(-.2,1.2,150),np.linspace(-.2,1.2,150))
grid=np.c_[gx.ravel(),gy.ravel()]; zz=sig(sig(grid@W1+b1)@W2+b2).reshape(gx.shape)
plt.contourf(gx,gy,zz,levels=20,cmap="coolwarm"); plt.scatter(X[:,0],X[:,1],c=y.ravel(),edgecolor="k")
plt.title("Two-layer XOR decision surface"); plt.show()
# Numerical gradient check for W1[0,0]
def loss_fn():
    return np.mean((sig(sig(X@W1+b1)@W2+b2)-y)**2)
eps=1e-5; old=W1[0,0]; W1[0,0]=old+eps; lp=loss_fn(); W1[0,0]=old-eps; lm=loss_fn(); W1[0,0]=old
print("Numerical gradient W1[0,0]:",(lp-lm)/(2*eps))
```
