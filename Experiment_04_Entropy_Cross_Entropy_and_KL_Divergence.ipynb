## Experiment 4 – Entropy, Cross-Entropy and KL Divergence

```python
from scipy.special import rel_entr
p=np.array([.5,.3,.2]) # observed Sunny, Cloudy, Rainy distribution
models={"Model A":np.array([.6,.25,.15]),"Model B":np.array([.3,.4,.3])}
H=stats.entropy(p,base=2)
print("Observed entropy:",H,"bits")
for name,q in models.items():
    ce=-np.sum(p*np.log2(q)); kl=stats.entropy(p,q,base=2)
    print(name,"Cross-entropy:",ce,"KL(P||Q):",kl,"H+KL:",H+kl)
q=models["Model A"]
print("KL(P||Q):",stats.entropy(p,q),"KL(Q||P):",stats.entropy(q,p))
y=np.array([0,1])
prob=np.linspace(.001,.999,200)
for actual in [0,1]:
    loss=-(actual*np.log(prob)+(1-actual)*np.log(1-prob))
    plt.plot(prob,loss,label=f"True label={actual}")
plt.xlabel("Predicted probability of class 1"); plt.ylabel("Binary cross-entropy"); plt.legend(); plt.grid(); plt.show()
```
