## Experiment 14 – Unfolded RNN Architecture Visualization

```python
import tensorflow as tf, networkx as nx
# 5-step, 3-feature sequence and 3-unit SimpleRNN
tf.random.set_seed(14)
rnn=tf.keras.layers.SimpleRNN(3,return_sequences=True,return_state=True)
seq=np.random.rand(1,5,3).astype("float32")
out,state=rnn(seq)
Wx,Wh,b=rnn.get_weights()
h=np.zeros(3); states=[]
for xt in seq[0]:
    h=np.tanh(xt@Wx+h@Wh+b); states.append(h.copy())
print("Manual hidden states:\n",np.round(states,4))
print("Keras hidden states:\n",np.round(out.numpy()[0],4))
G=nx.DiGraph()
for t in range(1,6):
    G.add_node(f"x{t}",layer=0); G.add_node(f"h{t}",layer=1); G.add_node(f"y{t}",layer=2)
    G.add_edge(f"x{t}",f"h{t}",label="W_xh"); G.add_edge(f"h{t}",f"y{t}",label="W_hy")
    if t>1:G.add_edge(f"h{t-1}",f"h{t}",label="W_hh")
pos={}
for t in range(1,6): pos[f"x{t}"]=(t,0); pos[f"h{t}"]=(t,1); pos[f"y{t}"]=(t,2)
nx.draw(G,pos,with_labels=True,node_size=900,node_color="lightblue",arrows=True)
nx.draw_networkx_edge_labels(G,pos,edge_labels=nx.get_edge_attributes(G,"label"),font_size=7)
plt.title("Unfolded RNN"); plt.show()
plt.plot(np.array(states)); plt.xlabel("Time step"); plt.ylabel("Hidden activation"); plt.title("Hidden states"); plt.show()
```
