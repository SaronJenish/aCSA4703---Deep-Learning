## Experiment 19 – PCA and Explained Variance Analysis

```python
from sklearn.datasets import load_digits
from sklearn.decomposition import PCA
digits=load_digits(); X=digits.data; Xc=X-X.mean(0)
cov=np.cov(Xc,rowvar=False); vals,vecs=np.linalg.eigh(cov); order=np.argsort(vals)[::-1]
vals=vals[order]; vecs=vecs[:,order]; ratio=vals/vals.sum(); cumulative=np.cumsum(ratio)
k=np.searchsorted(cumulative,.95)+1
print("Components for 95% variance:",k)
pca=PCA().fit(X); print("Sklearn components for 95%:",np.searchsorted(np.cumsum(pca.explained_variance_ratio_),.95)+1)
plt.plot(np.arange(1,len(ratio)+1),ratio,label="Individual"); plt.plot(np.arange(1,len(ratio)+1),cumulative,label="Cumulative")
plt.axhline(.95,color="red",ls="--"); plt.legend(); plt.xlabel("Components"); plt.show()
fig,axs=plt.subplots(1,5,figsize=(12,3))
for ax,i in zip(axs,range(5)): ax.imshow(vecs[:,i].reshape(8,8),cmap="gray"); ax.set_title(f"Eigenface {i+1}"); ax.axis("off")
plt.show()
for ncomp in [10,50,100]:
    pp=PCA(n_components=ncomp); rec=pp.inverse_transform(pp.fit_transform(X[:1]))
    plt.figure(); plt.subplot(1,2,1); plt.imshow(X[0].reshape(8,8),cmap="gray"); plt.title("Original")
    plt.subplot(1,2,2); plt.imshow(rec[0].reshape(8,8),cmap="gray"); plt.title(f"Reconstruction {ncomp}"); plt.show()
```
