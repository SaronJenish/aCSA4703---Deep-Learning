```python
# Common setup – run this cell first
import os, time, warnings, math
warnings.filterwarnings("ignore")
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             confusion_matrix, mean_absolute_error, mean_squared_error)
np.random.seed(42)
plt.rcParams["figure.figsize"] = (7, 4)
print("Setup complete")
```

## Experiment 18 – RNN, LSTM and GRU Hyperparameter Experimentation

```python
# IMDB sentiment dataset; small subset for quicker Colab execution
import tensorflow as tf, time
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.datasets import imdb
(xtr,ytr),(xte,yte)=imdb.load_data(num_words=8000)
xtr=pad_sequences(xtr[:8000],maxlen=120); ytr=ytr[:8000]
xte=pad_sequences(xte[:2000],maxlen=120); yte=yte[:2000]
rows=[]
for cell in ["SimpleRNN","LSTM","GRU"]:
    start=time.time()
    layer={"SimpleRNN":tf.keras.layers.SimpleRNN,"LSTM":tf.keras.layers.LSTM,"GRU":tf.keras.layers.GRU}[cell]
    model=tf.keras.Sequential([tf.keras.layers.Embedding(8000,32,input_length=120),layer(32),tf.keras.layers.Dropout(.3),tf.keras.layers.Dense(1,activation="sigmoid")])
    model.compile("adam","binary_crossentropy",metrics=["accuracy"])
    model.fit(xtr,ytr,epochs=2,batch_size=128,validation_split=.1,verbose=0)
    loss,acc=model.evaluate(xte,yte,verbose=0)
    rows.append([cell,acc,model.count_params(),time.time()-start])
print(pd.DataFrame(rows,columns=["Cell","Test accuracy","Parameters","Seconds"]))
```
