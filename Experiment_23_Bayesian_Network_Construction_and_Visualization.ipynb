## Experiment 23 – Bayesian Network Construction and Visualization

```python
# Install pgmpy if needed in Colab
%pip -q install pgmpy
import networkx as nx
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.factors.discrete import TabularCPD
from pgmpy.inference import VariableElimination
edges=[("Smoking","LungDisease"),("Pollution","LungDisease"),("LungDisease","Cough"),("LungDisease","Xray"),("LungDisease","Fatigue")]
model=DiscreteBayesianNetwork(edges)
cpds=[
 TabularCPD("Smoking",2,[[.7],[.3]]),
 TabularCPD("Pollution",2,[[.8],[.2]]),
 TabularCPD("LungDisease",2,[[.99,.7,.8,.1],[.01,.3,.2,.9]],evidence=["Smoking","Pollution"],evidence_card=[2,2]),
 TabularCPD("Cough",2,[[.8,.2],[.2,.8]],evidence=["LungDisease"],evidence_card=[2]),
 TabularCPD("Xray",2,[[.95,.1],[.05,.9]],evidence=["LungDisease"],evidence_card=[2]),
 TabularCPD("Fatigue",2,[[.7,.3],[.3,.7]],evidence=["LungDisease"],evidence_card=[2])]
model.add_cpds(*cpds); print("Model valid:",model.check_model())
G=nx.DiGraph(edges); pos=nx.spring_layout(G,seed=4)
nx.draw(G,pos,with_labels=True,node_color="lightblue",node_size=2200,arrows=True); plt.title("Medical Bayesian Network"); plt.show()
print("Active trail from Smoking:",model.active_trail_nodes("Smoking"))
```
