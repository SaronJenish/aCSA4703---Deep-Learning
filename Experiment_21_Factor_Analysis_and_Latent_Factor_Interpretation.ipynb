```python
# Common setup – run this cell first
import os, time, warnings, math
warnings.filterwarnings("ignore")
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             confusion_matrix, mean_absolute_error, mean_squared_error)
np.random.seed(42)
plt.rcParams["figure.figsize"] = (7, 4)
print("Setup complete")
```

## Experiment 21 – Factor Analysis and Latent Factor Interpretation

```python
from sklearn.decomposition import FactorAnalysis, PCA
# 300 employee responses, 10 Likert questions generated from latent factors
rng=np.random.default_rng(21); latent=rng.normal(size=(300,3))
load=np.array([[.9,.1,.1],[.8,.2,.1],[.7,.2,.2],[.1,.9,.1],[.2,.8,.2],[.1,.7,.2],[.1,.2,.9],[.2,.1,.8],[.2,.2,.7],[.4,.3,.4]])
raw=latent@load.T+rng.normal(0,.5,(300,10))
survey=np.clip(np.rint(3+raw),1,5)
cols=[f"Q{i}" for i in range(1,11)]; df=pd.DataFrame(survey,columns=cols)
print(df.describe()); print("Correlation:\n",df.corr().round(2))
for k in [2,3,4]:
    fa=FactorAnalysis(n_components=k,random_state=42).fit(df)
    print(f"\n{k} factors loadings:\n",pd.DataFrame(fa.components_.T,index=cols).round(2))
fa=FactorAnalysis(n_components=3,random_state=42).fit(df)
plt.imshow(fa.components_.T,cmap="coolwarm",aspect="auto"); plt.colorbar(); plt.yticks(range(10),cols); plt.xlabel("Factors"); plt.title("Factor loadings heatmap"); plt.show()
pca=PCA(n_components=3).fit(df)
print("FactorAnalysis noise variance:",fa.noise_variance_,"PCA explained variance:",pca.explained_variance_ratio_)
```
