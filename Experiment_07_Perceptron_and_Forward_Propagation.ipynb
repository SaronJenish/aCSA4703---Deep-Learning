## Experiment 7 – Perceptron and Forward Propagation

```python
# Perceptron learning rule on linearly separable synthetic loan data
rng=np.random.default_rng(7)
X=np.vstack([rng.normal([2,2],.5,(50,2)),rng.normal([4,4],.5,(50,2))])
y=np.r_[np.zeros(50),np.ones(50)]
w=np.zeros(2); b=0.; lr=.1; updates=[]
for epoch in range(30):
    errors=0
    for xi,yi in zip(X,y):
        pred=int(w@xi+b>=0); e=yi-pred
        w+=lr*e*xi; b+=lr*e; errors+=abs(e)
    updates.append(errors)
print("Weights:",w,"Bias:",b,"Accuracy:",accuracy_score(y,(X@w+b>=0).astype(int)))
plt.plot(updates); plt.xlabel("Epoch"); plt.ylabel("Misclassifications"); plt.show()
xx=np.linspace(X[:,0].min(),X[:,0].max(),100)
plt.scatter(X[:,0],X[:,1],c=y); 
if abs(w[1])>1e-8: plt.plot(xx,-(w[0]*xx+b)/w[1],"r")
plt.title("Perceptron decision boundary"); plt.show()
# XOR cannot be separated by one linear perceptron
Xxor=np.array([[0,0],[0,1],[1,0],[1,1]]); yxor=np.array([0,1,1,0])
ww=np.zeros(2); bb=0
for _ in range(100):
    for xi,yi in zip(Xxor,yxor):
        e=yi-int(ww@xi+bb>=0); ww+=.1*e*xi; bb+=.1*e
print("XOR predictions:",(Xxor@ww+bb>=0).astype(int),"Expected:",yxor)
z=np.linspace(-5,5,100)
plt.plot(z,(z>=0).astype(float),label="Step"); plt.plot(z,1/(1+np.exp(-z)),label="Sigmoid"); plt.plot(z,np.maximum(0,z),label="ReLU")
plt.legend(); plt.title("Activation functions"); plt.show()
```
