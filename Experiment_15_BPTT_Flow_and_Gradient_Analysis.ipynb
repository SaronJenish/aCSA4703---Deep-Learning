## Experiment 15 – BPTT Flow and Gradient Analysis

```python
import tensorflow as tf
# Compare gradient propagation over sequence lengths and activations
tf.random.set_seed(15)
def gradient_profile(length,activation):
    layer=tf.keras.layers.SimpleRNN(16,activation=activation,return_sequences=True)
    x=tf.Variable(tf.random.normal((1,length,8)))
    with tf.GradientTape() as tape:
        y=layer(x); loss=tf.reduce_sum(y[:,-1,:])
    g=tape.gradient(loss,x).numpy()[0]
    return np.linalg.norm(g,axis=1)
fig,ax=plt.subplots()
for L in [10,50,100]:
    for act in ["tanh","relu"]:
        g=gradient_profile(L,act)
        ax.plot(np.arange(L),g,label=f"{act}, L={L}")
ax.set_yscale("log"); ax.set_xlabel("Input time step"); ax.set_ylabel("Gradient norm"); ax.legend(); ax.set_title("BPTT gradient magnitude"); plt.show()
```
