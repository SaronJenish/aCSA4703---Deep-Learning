```python
# Common setup – run this cell first
import os, time, warnings, math
warnings.filterwarnings("ignore")
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             confusion_matrix, mean_absolute_error, mean_squared_error)
np.random.seed(42)
plt.rcParams["figure.figsize"] = (7, 4)
print("Setup complete")
```

## Experiment 9 – Multilayer Perceptron Implementation

```python
# Synthetic customer churn dataset
import tensorflow as tf
rng=np.random.default_rng(9); n=1200
tenure=rng.integers(1,72,n); charges=rng.normal(70,20,n).clip(20,150)
contract=rng.integers(0,3,n); complaints=rng.poisson(1,n); usage=rng.normal(30,10,n).clip(1,80)
logit=-1.5-.035*tenure+.018*charges+.65*(contract==0)+.45*complaints
prob=1/(1+np.exp(-logit)); churn=(rng.random(n)<prob).astype(int)
df=pd.DataFrame({"tenure":tenure,"charges":charges,"contract":contract,"complaints":complaints,"usage":usage})
X=pd.get_dummies(df,columns=["contract"]).astype(float)
X=StandardScaler().fit_transform(X); Xtr,Xte,ytr,yte=train_test_split(X,churn,test_size=.2,random_state=42,stratify=churn)
model=tf.keras.Sequential([tf.keras.layers.Input((X.shape[1],)),tf.keras.layers.Dense(32,activation="relu"),tf.keras.layers.Dense(16,activation="relu"),tf.keras.layers.Dense(1,activation="sigmoid")])
model.compile(optimizer="adam",loss="binary_crossentropy",metrics=["accuracy"])
h=model.fit(Xtr,ytr,epochs=20,batch_size=32,validation_split=.2,verbose=0)
pred=(model.predict(Xte,verbose=0).ravel()>.5).astype(int)
print("Accuracy:",accuracy_score(yte,pred),"Precision:",precision_score(yte,pred,zero_division=0),"Recall:",recall_score(yte,pred,zero_division=0),"F1:",f1_score(yte,pred,zero_division=0))
print("Confusion matrix:\n",confusion_matrix(yte,pred))
plt.plot(h.history["accuracy"],label="train"); plt.plot(h.history["val_accuracy"],label="validation"); plt.legend(); plt.title("MLP accuracy"); plt.show()
```
