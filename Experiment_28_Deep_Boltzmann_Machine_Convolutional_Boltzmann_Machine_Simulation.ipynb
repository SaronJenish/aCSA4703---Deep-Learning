## Experiment 28 – Deep Boltzmann Machine / Convolutional Boltzmann Machine Simulation

```python
# Two-hidden-layer mean-field DBM-style simulation on small binary image patches
from sklearn.datasets import load_digits
digits=load_digits(); X=(digits.data/16>.5).astype(float)
rng=np.random.default_rng(28); W1=rng.normal(0,.03,(64,48)); W2=rng.normal(0,.03,(48,24))
b1=np.zeros(48); b2=np.zeros(24); vb=np.zeros(64)
def sigmoid(z): return 1/(1+np.exp(-np.clip(z,-30,30)))
h1=np.zeros((len(X),48)); h2=np.zeros((len(X),24))
for epoch in range(20):
    # Mean-field updates for approximate hidden posterior
    for _ in range(5):
        h1=sigmoid(X@W1+h2@W2.T+b1)
        h2=sigmoid(h1@W2+b2)
    recon=sigmoid(h1@W1.T+vb)
    # Simple reconstruction-driven update for demonstration
    err=X-recon; W1+=.02*(err.T@h1)/len(X); vb+=.02*err.mean(0)
print("DBM-style reconstruction MSE:",np.mean((X-recon)**2))
fig,axs=plt.subplots(2,5,figsize=(10,4))
for i,ax in enumerate(axs.flat):
    ax.imshow(recon[i].reshape(8,8),cmap="gray"); ax.axis("off")
plt.suptitle("Reconstructed digit patches"); plt.show()
plt.imshow(W1.T[:16].reshape(16,8,8)[0],cmap="gray"); plt.title("Example learned filter"); plt.show()
```
