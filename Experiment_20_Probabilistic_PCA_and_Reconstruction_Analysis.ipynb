## Experiment 20 – Probabilistic PCA and Reconstruction Analysis

```python
from sklearn.decomposition import PCA
rng=np.random.default_rng(20); Z=rng.normal(size=(500,2)); W=rng.normal(size=(2,5))
X=Z@W+rng.normal(0,.4,size=(500,5))
X[::30,2]=np.nan
# Simple mean imputation for missing sensor readings
X=np.where(np.isnan(X),np.nanmean(X,axis=0),X)
sc=StandardScaler(); Xs=sc.fit_transform(X)
errors=[]; models=[]
for k in range(1,5):
    pca=PCA(n_components=k).fit(Xs); rec=pca.inverse_transform(pca.transform(Xs))
    errors.append(np.mean((Xs-rec)**2)); models.append(pca)
print("Estimated noise variance (4 components):",models[-1].noise_variance_)
plt.plot(range(1,5),errors,"o-"); plt.xlabel("Components"); plt.ylabel("Reconstruction MSE"); plt.show()
pca=models[1]; znew=np.random.normal(size=(5,2)); generated=znew@pca.components_[:2]+pca.mean_
print("Example generated latent sensor samples:\n",generated)
```
