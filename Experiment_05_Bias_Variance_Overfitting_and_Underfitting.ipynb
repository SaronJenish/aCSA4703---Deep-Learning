```python
# Common setup – run this cell first
import os, time, warnings, math
warnings.filterwarnings("ignore")
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             confusion_matrix, mean_absolute_error, mean_squared_error)
np.random.seed(42)
plt.rcParams["figure.figsize"] = (7, 4)
print("Setup complete")
```

## Experiment 5 – Bias–Variance, Overfitting and Underfitting

```python
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
from sklearn.pipeline import make_pipeline
rng=np.random.default_rng(42)
X=(rng.random(150)*10).reshape(-1,1); y=2+1.5*X[:,0]+.8*np.sin(X[:,0]*1.5)+rng.normal(0,1,150)
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.25,random_state=42)
degrees=[1,3,9,15]; train_err=[]; val_err=[]
xx=np.linspace(0,10,300).reshape(-1,1)
fig,axs=plt.subplots(2,2,figsize=(12,8))
for d,ax in zip(degrees,axs.flat):
    model=make_pipeline(PolynomialFeatures(d),LinearRegression()).fit(Xtr,ytr)
    train_err.append(mean_squared_error(ytr,model.predict(Xtr)))
    val_err.append(mean_squared_error(yte,model.predict(Xte)))
    ax.scatter(Xtr,ytr,s=12); ax.plot(xx,model.predict(xx),color="red"); ax.set_title(f"Polynomial degree {d}")
plt.tight_layout(); plt.show()
plt.plot(degrees,train_err,"o-",label="Training MSE"); plt.plot(degrees,val_err,"o-",label="Validation MSE")
plt.xlabel("Polynomial degree"); plt.ylabel("MSE"); plt.legend(); plt.grid(); plt.show()
# Bootstrap approximation of prediction variance and squared bias
preds=[]
for _ in range(40):
    ids=rng.integers(0,len(Xtr),len(Xtr))
    m=make_pipeline(PolynomialFeatures(3),LinearRegression()).fit(Xtr[ids],ytr[ids])
    preds.append(m.predict(xx))
preds=np.array(preds); true=2+1.5*xx[:,0]+.8*np.sin(xx[:,0]*1.5)
print("Approx. degree-3 bias^2:",np.mean((preds.mean(0)-true)**2))
print("Approx. degree-3 variance:",np.mean(preds.var(0)))
print("Best validation degree:",degrees[int(np.argmin(val_err))])
```
